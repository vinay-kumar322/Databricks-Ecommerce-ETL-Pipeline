```python
%sql
USE CATALOG data_analyst_demo;
USE SCHEMA ecommerce;
```

```python
%sql
SELECT
ROUND(SUM(payment_value),2) AS Total_Revenue
FROM gold_sales;
```

```python
%sql
SELECT
COUNT(DISTINCT order_id) AS Total_Orders
FROM gold_sales;
```

```python
%sql
SELECT
COUNT(DISTINCT customer_unique_id) AS Total_Customers
FROM gold_sales;
```

```python
%sql
SELECT
product_id,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY product_id
ORDER BY Revenue DESC
LIMIT 10;
```

```python
%sql
SELECT
seller_id,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY seller_id
ORDER BY Revenue DESC
LIMIT 10;
```

```python
%sql
SELECT
state,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY state
ORDER BY Revenue DESC;
```

```python
%sql
SELECT
payment_type,
COUNT(*) AS Total_Payments,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY payment_type
ORDER BY Revenue DESC;
```

```python
%sql
SELECT
AVG(review_score) AS Average_Rating
FROM silver_reviews;
```

```python
%sql
SELECT
review_score,
COUNT(*) AS Total_Reviews
FROM silver_reviews
GROUP BY review_score
ORDER BY review_score;
```

```python
%sql
SELECT
YEAR(order_purchase_timestamp) AS Year,
MONTH(order_purchase_timestamp) AS Month,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY
YEAR(order_purchase_timestamp),
MONTH(order_purchase_timestamp)
ORDER BY Year,Month;
```

```python
%sql
SELECT
city,
ROUND(SUM(payment_value),2) AS Revenue
FROM gold_sales
GROUP BY city
ORDER BY Revenue DESC
LIMIT 10;
```

```python
%sql
SELECT
order_status,
COUNT(*) AS Orders
FROM gold_sales
GROUP BY order_status
ORDER BY Orders DESC;
```

```python
%sql
SELECT
AVG(
DATEDIFF(
order_delivered_customer_date,
order_purchase_timestamp
)
) AS Avg_Delivery_Days
FROM silver_orders
WHERE order_status='delivered';
```

```python
%sql
SELECT
customer_unique_id,
COUNT(DISTINCT order_id) AS Total_Orders
FROM gold_sales
GROUP BY customer_unique_id
ORDER BY Total_Orders DESC
LIMIT 10;
```
