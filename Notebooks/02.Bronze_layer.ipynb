```python
%sql
USE CATALOG data_analyst_demo;
USE SCHEMA ecommerce;
```


## Bronze_customers

```python
customers_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/data_analyst_demo/ecommerce/olist_data/olist_customers_dataset.csv")
)

customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_customers")
```

```python
%sql
SELECT *
FROM bronze_customers
LIMIT 10;
```

## Bronze_reviews

```python
%sql
DROP TABLE IF EXISTS data_analyst_demo.ecommerce.bronze_reviews;
```

```python
reviews_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .option("multiLine", True)
         .option("quote", "\"")
         .option("escape", "\"")
         .csv("/Volumes/data_analyst_demo/ecommerce/olist_data/olist_order_reviews_dataset.csv")
)
```

```python
reviews_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("data_analyst_demo.ecommerce.bronze_reviews")
```
